# دائرة التشابك الكمومي بين السطح والباطن (Bell State)

هذا الدفتر يبني دائرة كمومية بلغة Qiskit لإنشاء حالة بل (Bell State) بين كيوبرتين:
- **الكيوبرت 0**: مستشعر السطح
- **الكيوبرت 1**: أثير باطن الأرض

ثم يُشغّله على المحاكي المحلي `AerSimulator` لـ 1024 محاولة (Shots) ويعرض النتائج كمخطط بياني (Histogram).

```python
# استدعاء الحزم البرمجية الأساسية من Qiskit
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

# 1. إنشاء الدائرة الكمومية: 2 كيوبرت (Qubits) و2 بت كلاسيكي للقراءة
# الكيوبرت 0 = مستشعر السطح | الكيوبرت 1 = أثير باطن الأرض
qc = QuantumCircuit(2, 2)

# 2. حقن بوابة هادامارد (H) على الكيوبرت 0 لوضعه في حالة التراكب (موجة احتمالية)
qc.h(0)

# 3. حقن بوابة التحكم بالنفي (CNOT / CX) لربط الكيوبرت 0 بالكيوبرت 1 (صناعة التشابك)
qc.cx(0, 1)

# 4. إضافة أمر القياس لقراءة النتائج من الكيوبرت وتحويلها لبتات كلاسيكية
qc.measure([0, 1], [0, 1])

# رسم الدائرة بشكل نصي للتأكد من الترتيب الصحيح
print("شكل الدائرة الكمومية المحاكة:")
print(qc.draw())
```

```python
# إعداد المحاكي المحلي (Simulator)
simulator = AerSimulator()

# تحويل الدائرة الرياضية لتتوافق مع معايير المحاكي
compiled_circuit = transpile(qc, simulator)

# إطلاق أمر التشغيل والمحاكاة لـ 1024 مرة (Shots) لالتقاط النسب الإحصائية
job = simulator.run(compiled_circuit, shots=1024)

# استخراج وقراءة النتائج والترددات المشتبكة
result = job.result()
counts = result.get_counts()

print("\nنتائج قياس التردد المشتبك بين السطح والباطن:")
print(counts)
```

```python
# رسم النتائج كمخطط بياني (Histogram) لتصوّر توزيع الحالات
from qiskit.visualization import plot_histogram

print("المخطط البياني لتوزيع النتائج:")
plot_histogram(counts)
```
